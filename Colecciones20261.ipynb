# Introducción de Listas, tuplas, conjuntos  y diccionarios

## listas (list)

1. Se crear con []
2. son mutables (de pueden modificar)
3. Pueden tener elemenos distintos

```python
Frutas=["Manzanas", "Peras", "Naranjas"]
print("Lista de mercado:", Frutas)
```

```python
Frutas.append("Fresas")# agregar elemento
print("Lista de mercado:", Frutas)
```

```python
Frutas.remove("Peras")# eliminar elemento
print("Lista de mercado:", Frutas)
```

```python
Frutas[0]="Uva" #modificar
print("Lista de mercado:", Frutas)
```

```python
Calificación=[1,5,7,10,6,9,8,11,12]
#promedio
print("Promedio:", round(sum(Calificación)/len(Calificación),2))
```

```python
#Filtro >7 aprueben
aprobados=[nota for nota in
           Calificación if nota>7]
print("Aprobados:", aprobados)
```

# Tuple

1. se crean con ()
2. son inmutables (no se pueden modificar después la creación)
3. Se usan para datos fijos


```python
Coordenadas=(10,20)
print("Tuple", Coordenadas)
print("Coordenada X", Coordenadas[0])
print("Coordenada Y", Coordenadas[1])
```

```python
#Sensores de ubicación (Latitud, Longitud)
Sensores=((4.6,-74.08),
          (34.05, -118.12),
          (48.78, 2.56))

for lat, lon in Sensores:
  print(f"Latitud: {lat}, Longitud: {lon}")
```

# Conjuntos (set)

1. Se crean con {} o con set()
2. No tienen orden
3. no permite duplicados
4. se usan para operaciones de teoria de conjuntos: unión, intersección y deferente.

```python
# x|x 1<=x<=6
A={1,2,3,4,5,6}
print(A)
# x|x 4<=x<=10
B={4,5,6,7,8,9,10}
print(B)
```

```python
print("Unión", A|B)#unión
print("Intersección", A&B)#intercción
print("Diferencia", A-B)#diferencia
```

```python
Unión=(A|B)
Unión
```

```python
Ciudades=["Bogotá","Bogotá","Cali", "Medellin", "Tunja", "Cartagena", "Santa Marta"]

Ciudad=set(Ciudades)# no existe duplicados
print(Ciudad)
```

# Diccionario

1. Crear una clave: Valor.
2. Permite asociar una calve con valor.
3. Muy usados en ciencias de datos.
4. se crean con {}
5. Se puede hacer modificaciones en el diccionario después de crear


```python
Estudiante={"Nombre":"Manuel",
            "Edad": 28,
            "Carrera": "Ingeniería"}
print(Estudiante)
```

```python
Estudiante["Edad"]=22
print(Estudiante)
```

```python
Estudiantes=[]

for i in range(3):
  nombre=input("Nombre: ")
  Edad= int(input("Edad: "))
  carrera= input("Carrera: ")

  Estudiantes.append({"Nombre":nombre,
                      "Edad":Edad,
                      "Carrera":carrera})
print("\nLista de estudiantes ingresada")
for e in Estudiantes:
  print(e)
```

```python
Estudiantes = []

n = int(input("¿Cuántos estudiantes desea ingresar? "))

for i in range(n):
    print(f"\nEstudiante {i+1}")
    nombre = input("Nombre: ")
    edad = int(input("Edad: "))
    carrera = input("Carrera: ")

    Estudiantes.append({
        "Nombre": nombre,
        "Edad": edad,
        "Carrera": carrera
    })

print("\nLista de estudiantes ingresada:")

for e in Estudiantes:
    print(f"Nombre: {e['Nombre']}, Edad: {e['Edad']}, Carrera: {e['Carrera']}")
```

```python
import pandas as pd

df = pd.DataFrame(Estudiantes)
print(df)
```
