# **Series**

Una Series de Pandas es un arreglo (array) unidimensional de datos, asociado a un conjunto de etiquetas de índice que identifican cada elemento. En términos simples, se puede imaginar como una columna de Excel o una columna en una tabla, con sus valores y con un índice que actúa como etiqueta para cada fila. La definición formal según la documentación es un "arreglo unidimensional con etiquetas de eje",lo cual incluye la posibilidad de usar índices que representen fechas en el caso de series temporales.

## DataFrame

Un DataFrame es la estructura de datos más representativa de Pandas: se trata de una colección de Series organizadas en columnas, compartiendo un mismo índice de filas. Podemos pensarlo como una tabla bidimensional de datos, similar a una hoja de cálculo de Excel o a una tabla en SQL, donde cada columna tiene un nombre (etiqueta de columna) y cada fila está identificada por un índice. En un DataFrame, cada columna es en realidad una Series de Pandas, lo que implica que todos los valores de una columna son del mismo tipo (númerico, cadena, booleano, etc.), mientras que las distintas columnas pueden ser de tipos diferentes entre sí.

# **¿Qué es un Panel en este contexto?**

En términos prácticos, un Panel es un array de tres dimensiones (3D), también conocido como cubo de datos. Es útil para representar datos que varían según tres ejes, como:

* Clientes × Meses × Variables

* Tiendas × Días × Productos

* Regiones × Años × Indicadores económicos

```python
np.random.seed(9876)#semilla

# Dimensiones: 3 clientes × 4 meses × 2 variables
panel = np.random.uniform(2000, 7000, size=(3, 4, 2)) # 3 bloques(matrices), 4 filas, 2 columnas

print("Array 3D (Panel de ingresos):")
print(panel)
```

```python
# Crear índices y etiquetas
clientes = ['Cliente A', 'Cliente B', 'Cliente C'] #Bloque
meses = ['Enero', 'Febrero', 'Marzo', 'Abril'] # filas
variables = ['Ingreso Bruto', 'Ingreso Neto'] #columnas

# Convertir a DataFrame MultiIndex
panel_df = pd.DataFrame(
    panel.reshape(12, 2), # 12=3*4 bloque*filas, 2 columnas
    index=pd.MultiIndex.from_product([clientes, meses], names=['Cliente', 'Mes']),
    columns=variables
)

print("\nPanel convertido a tabla con MultiIndex:")
print(panel_df)
```

| Operación                  | Código                        | Dimensión impactada        |
| -------------------------- | ----------------------------- | -------------------------- |
| Media por cliente          | `np.mean(panel, axis=1)`      | Meses                      |
| Media por variable         | `np.mean(panel, axis=(0,1))`  | Clientes y meses           |
| Suma por variable (global) | `np.sum(panel[:, :, 1])`      | Variable específica        |
| Máximo por mes y variable  | `np.max(panel, axis=0)`       | Clientes                   |
| Reducción bruto - neto     | `panel[:,:,0] - panel[:,:,1]` | Diferencia entre variables |


```python
np.mean(panel_df, axis=1)  # Promedia los 4 meses para cada cliente
```

```python
panel_df.mean()  # Promedia todos los clientes y meses
```

```python
np.mean(panel[:, :, 1]) #Promedio de ingreso neto (variable 1):
```

## Suma total por variable

```python
## Suma total de ingresos brutos (variable 0):
np.sum(panel[:, :, 0])
```

```python
# Suma por cliente y mes (matriz 3x4):
np.sum(panel_df, axis=0)  # Suma bruto + neto
```

## Comparaciones y máscaras booleanas

```python
## Identificar ingresos brutos > 6000:
panel[:, :, 0] > 6000
```

```python
# Filtrar solo los ingresos netos menores a 3000:
panel[:, :, 1][panel[:, :, 1] < 3000]
```

## Reducción y agregación personalizada

```python
# Diferencia entre bruto y neto por cliente y mes:
diferencia = panel[:, :, 0] - panel[:, :, 1]
print(diferencia)
```

```python
# Coeficiente de variación por variable:
coef_var = np.std(panel, axis=(0, 1)) / np.mean(panel, axis=(0, 1))
print(coef_var)
```

## Transformaciones

```python
# Redondear valores
np.round(panel, 2)
```

```python
# Normalizar entre 0 y 1:
norm = (panel - np.min(panel)) / (np.max(panel) - np.min(panel))
print(norm)
```
