# Operadores Relacionales en Python



| Operador | Nombre        | Ejemplo en Python | Resultado |
| -------- | ------------- | ----------------- | --------- |
| `>`      | Mayor         | `5 > 3`           | `True`    |
| `>=`     | Mayor o igual | `5 >= 5`          | `True`    |
| `<`      | Menor         | `2 < 6`           | `True`    |
| `<=`     | Menor o igual | `4 <= 3`          | `False`   |
| `==`     | Igual         | `7 == 7`          | `True`    |
| `!=`     | Diferente     | `10 != 8`         | `True`    |


```python
# Lista de ejemplos
x = 5
y = 3

print("Operadores Relacionales en Python:\n")

# Mayor
print(f"{x} > {y}  ->", x > y)

# Mayor o igual
print(f"{x} >= {y} ->", x >= y)

# Menor
print(f"{x} < {y}  ->", x < y)

# Menor o igual
print(f"{x} <= {y} ->", x <= y)

# Igual
print(f"{x} == {y} ->", x == y)

# Diferente
print(f"{x} != {y} ->", x != y)
```

# **Condicionales: if, elif, else**

Los condicionales en Python son estructuras de control de flujo que permiten tomar decisiones en un programa.
Se usan cuando queremos que el código ejecute unas instrucciones solo si se cumple una condición (verdadero o falso).

##  **if**

Significa “si”.

Se ejecuta solo si la condición es verdadera (True).

Si la condición es falsa, no hace nada.

```python
edad = 20

if edad >= 18:
    print("Eres mayor de edad")
```

```python
a = 1
b = 2

if b > a:
    print(" b es mayor que a")
```

```python
compra = 120

if compra > 100:
    print("Aplica un 10% de descuento en la compra.")
```

## **else**

Significa “si no”.

Se ejecuta cuando la condición del if es falsa.

Solo puede haber un else en un bloque condicion

```python
nota = 5

if nota >= 6:
    print("Aprobado")
else:
    print("Reprobado")
```

```python
a = 1
b = 2

if a < b:
    print(" b es mayor que a")
else:
    print("a es mayor que b")
```

```python
saldo = -2500

if saldo >= 0:
    print("Cliente Estable")
else:
    print("Cliente en Riesgo Alto")
```

```python
temperatura = 32

if temperatura <= 30:
    print("Temperatura aceptable, lote en buen estado.")
else:
    print("Alerta: Temperatura fuera de rango, posible riesgo.")
```

```python
ingresos = 4200

if ingresos < 3000:
    print("Cliente de Ingresos Bajos")
else:
    print("Cliente de Ingresos Altos")
```

## 3. elif

Significa “si no, entonces si…”.

Se usa para comprobar múltiples condiciones de forma ordenada.

Puede haber varios elif en un bloque.

Se ejecuta solo la primera condición verdadera y el resto se ignoran.

```python
nota = 8

if nota < 6:
    print("Reprobado")
elif nota < 9:
    print("Aprobado")
else:
    print("Excelente")
```

```python
x = 1

if x > 10:
    print(" x es mayor que 10!")
elif x < 10:
      print("x es menor que 10!")
elif x < 20 :
      print("x es menor que 20!")
else:
     print("x es igual a 10")
```

```python
#Bajo: < 2000
#Medio: 2000–4000
#Alto: > 4000

ingresos = 3200

if ingresos < 2000: # Bajo: < 2000
    print("Cliente de Ingresos Bajos")
elif ingresos <= 4000: #Medio: 2000–4000
    print("Cliente de Ingresos Medios")
else: #Alto: > 4000
    print("Cliente de Ingresos Altos")
```

```python
#Baja: 1–2
#Media: 3
#Alta: 4–5

satisfaccion = 4

if satisfaccion <= 2:
    print("Satisfacción Baja")
elif satisfaccion == 3:
    print("Satisfacción Media")
else:
    print("Satisfacción Alta")
```

```python
#Fría: < 20°C
#Normal: 20°C–30°C
#Crítica: > 30°C

temperatura = 28.5

if temperatura < 20:
    print("Temperatura Fría")
elif temperatura <= 30:
    print("Temperatura Normal")
else:
    print("Temperatura Crítica")

```

En una empresa de retail, el equipo de analítica tiene los ingresos mensuales de un grupo de clientes y desea segmentarlos en tres categorías:

* Bajo: ingresos < 2000

* Medio: ingresos entre 2000 y 4000

* Alto: ingresos > 400

```python
# Lista de clientes con sus ingresos
clientes = {
    "C101": 1500,
    "C102": 2500,
    "C103": 4200,
    "C104": 3800,
    "C105": 1800
}

# Clasificación de ingresos usando for + if ... elif ... else
for cliente, ingreso in clientes.items():
    if ingreso < 2000:
        categoria = "Bajo"
    elif ingreso <= 4000:
        categoria = "Medio"
    else:
        categoria = "Alto"

    print(f"Cliente {cliente} → Ingreso: ${ingreso} → Categoría: {categoria}")
```

Control de calidad: clasificar temperaturas de sensores

```python
temperaturas = [19.5, 22.1, 29.9, 30.2, 27.8, 18.9, 31.4]

for t in temperaturas:
    if t < 20:
        estado = "Fría"
    elif t <= 30:
        estado = "Normal"
    else:
        estado = "Crítica"
    print(f"{t:.1f} °C → {estado}")
```

Etiquetar rendimiento de modelos por accuracy

```python
accuracies = {
    "modelo_A": 0.78, "modelo_B": 0.83, "modelo_C": 0.91, "modelo_D": 0.66
}

for nombre, acc in accuracies.items():
    if acc < 0.75:
        etiqueta = "Por mejorar"
    elif acc < 0.90:
        etiqueta = "Bueno"
    else:
        etiqueta = "Excelente"
    print(f"{nombre} → accuracy={acc:.2f} → {etiqueta}")
```

Clasificar probabilidad de churn (bajo/medio/alto)

```python
prob_churn = {
    "u001": 0.12, "u002": 0.31, "u003": 0.67, "u004": 0.49, "u005": 0.85
}

for user, p in prob_churn.items():
    if p < 0.3:
        nivel = "Bajo"
    elif p <= 0.6:
        nivel = "Medio"
    else:
        nivel = "Alto"
    print(f"Usuario {user} → p_churn={p:.2f} → Riesgo: {nivel}")
```

# **Estructuras de Control en Python – Bucles**

El bucle for se usa para recorrer elementos de una secuencia (listas, cadenas, rangos, etc.).

```python
# lista
frutas = ["manzana", "pera", "uva"]

for fruta in frutas:
    print("Me gusta la", fruta)
```

```python
#range
for i in range(1, 6):
    print("Número:", i)

```

## while

El bucle while repite un bloque mientras una condición sea verdadera.

```python
contador = 1

while contador <= 5:
    print("Contador:", contador)
    contador += 1
```

Contador regresivo (Cuenta atrás)

```python
contador = 5

while contador > 0:
    print("Cuenta regresiva:", contador)
    contador -= 1

print("¡Despegue! ")
```

Validar entrada de usuario

```python
password = "python123"
entrada = ""

while entrada != password:
    entrada = input("Introduce la contraseña: ")

print("Acceso concedido ✅")
```

Sumar hasta que el usuario diga "0"

```python
suma = 0
numero = int(input("Ingresa un número (0 para salir): "))

while numero != 0:
    suma += numero
    numero = int(input("Ingresa otro número (0 para salir): "))

print("La suma total es:", suma)
```

## break

El comando break interrumpe el bucle por completo, incluso si no ha terminado.

```python
for num in range(1, 10):
    if num == 5:
        break
    print(num)
```

Buscar un número en una lista

```python
numeros = [3, 7, 12, 19, 25, 30]

for n in numeros:
    if n == 19:
        print("Número encontrado:", n)
        break
```

Detenerse al encontrar una palabra prohibida

```python
palabras = ["hola", "python", "curso", "prohibido", "final"]

for palabra in palabras:
    if palabra == "prohibido":
        print("Se encontró una palabra prohibida. Deteniendo...")
        break
    print("Procesando:", palabra)
```

Juego de intentos con while

```python
intentos = 0
clave_secreta = "python"
while True:
    entrada = input("Introduce la clave: ")
    intentos += 1

    if entrada == clave_secreta:
        print("Acceso concedido en", intentos, "intentos ")
        break
    else:
        print("Clave incorrecta, intenta de nuevo.")
```

## continue

El comando continue salta a la siguiente iteración, sin ejecutar lo que sigue en el ciclo actual.

```python
for num in range(1, 6):
    if num == 3:
        continue
    print(num)
```

Saltar números pares

```python
for i in range(1, 10):
    if i % 2 == 0:
        continue  # se salta los pares
    print("Número impar:", i)
```

Saltar palabra específica en una lista

```python
palabras = ["python", "java", "c++", "javascript"]

for p in palabras:
    if p == "c++":
        continue  # ignora c++
    print("Lenguaje:", p)
```

Ignorar entradas vacías de usuario



```python
while True:
    texto = input("Escribe algo (o 'salir' para terminar): ")

    if texto == "":
        continue  # salta entradas vacías

    if texto.lower() == "salir":
        print("Programa terminado ✅")
        break

    print("Ingresaste:", texto)
```
