# Una lista donde se pueda utiliza las funciones de python



```python
Precio=[]
Precio.append(1200)
Precio.append(2000)
Precio.append(2340)
Precio.append(1250)
Precio.append(2350)
BasePrecio=Precio.copy()

#Eliminar valor
Precio.remove(2350)
Precio.insert(4,5630)
Precio.insert(1,2300)
Precio.pop()# elimina el ultimo de la lista
Precio.sort()# organiza los datos de la lista de mayor a menor
Precio.remove(1250)
Precio.insert(1,5000)
Precio.sort()
print(Precio)
print(BasePrecio)
```

# Diccionario

```python
#Biblioteca para un diccionario
Solicitudes=[1,2,4,5]
Bib={}
Bib["NombreLibro"]=Bib.copy()
Bib["Solicitudes"]=list(Solicitudes)
Bib["Edad"]=[24, 20, 18, 30]
print(Bib)
```

```python
Bib["Edad"].append(80)
Bib["Solicitudes"].append(4)
print(Bib)
```

```python
Bib["Edad"].pop()
Bib["Solicitudes"].pop()
print(Bib)
```

# filtros

```python
lista3=np.array([10,18,28,28,30,35,48,51,54,59,60,62,63,88])
rango=lista3[(lista3>28)]
print(rango)
```

```python
rango=lista3[(lista3>=28)]
print(rango)
```

```python
rango=lista3[(lista3<49)]
print(rango)
```

```python
rango=lista3[(lista3<=60)]
print(rango)
```

```python
# [ <= menor igual que
# ( < menor que
# [ >= mayor igual que
# ( > mayor que
rango=lista3[(lista3>=30) & (lista3<60)] # [30<= x < 60)
print(rango)
```

```python
rango=lista3[(lista3>30) & (lista3<60)]
print(rango)
```

# Loop for (ciclos for())

```python
# i la nueva variable para el ciclo
datos=np.arange(0,30)
print(datos)
```

```python
for i in datos:
  print(i)
```

```python
for i in datos:
  v=i**2
  print(v)
```

```python
for incremento in datos:
  IVA=incremento*0.12+incremento
  print(IVA)
```

```python
años=np.array([12,45,24,26,20,40,23,40])
print(años)
```

```python
for i in años:
  i=i+10
  print(i)
```

```python
for x in años:
  x=x*0.5
  print(x)
```
