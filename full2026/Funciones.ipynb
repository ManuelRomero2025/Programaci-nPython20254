La tabla muestra los tipos de datos enteros en NumPy, según su tamaño en bits, si son con signo (negativos y positivos) o sin signo (solo positivos), y el rango de valores que pueden almacenar.



| Tipo     | Bits | Rango de valores                           |
| -------- | ---- | ------------------------------------------ |
| `int8`   | 8    | -128 a 127                                 |
| `int16`  | 16   | -32,768 a 32,767                           |
| `int32`  | 32   | -2,147,483,648 a 2,147,483,647             |
| `int64`  | 64   | -9.2×10¹⁸ a 9.2×10¹⁸                       |
| `uint8`  | 8    | 0 a 255 (sin signo)                        |
| `uint16` | 16   | 0 a 65,535 (sin signo)                     |
| `uint32` | 32   | 0 a 4,294,967,295 (sin signo)              |
| `uint64` | 64   | 0 a 18,446,744,073,709,551,615 (sin signo) |

# tupla

una secuencia de datos de diferentes tipos.

* tienen orden
* puden contener elementos diferentes
* Son inmutables. no pueden cambiar los elementos.
* () parentesis abierto


```python
data1=(12,14,15,20,10,13,12,25,20, "Manuel", "Ginna")
print(data1[0])
print(data1[5])
print(data1[9])
print(data1[10])
```

```python
#data1[9]=100
#data1[10]=120
```

```python
vector=np.array([12,13,15,16,18])
vector[1]=20
print(vector)
```

```python
data2=([14,13,14,20,25,28,30,"Manuel"])
data2[7]=100
print(data2)
```

```python
data3=[(14,13,14,20,25,28,30,"Manuel")]
# Para acceder a un elemento dentro de la tupla que está en la lista:
# data3[0] accede a la tupla completa, y [1] accede al segundo elemento de esa tupla
print(data3[0][1])
print(data3[0][0])
#data3[0][7]=110
```

# Ingresar datos

```python
Ingresos=[]
Ingresos.append(1000)
Ingresos.append(1450)
Ingresos.append(2500)
#Remover o eliminar datos
Ingresos.remove(1000)
Ingresos.remove(2500)
#Egregar datos
Ingresos.append(4500)
Ingresos.append("Manuel")
```

```python
print(Ingresos)
```

```python
Gatos=([])
Gatos.append(1500)
Gatos.append(1400)
Gatos.append(1450)
Gatos.append(2450)
Gatos.append(2345)
#Gatos.remove(1900)
Gatos.insert(0,2500)
Gatos.insert(4,2800)
#
Gastos=Gatos.copy()
Gatos.insert(2,2800)
# Gatos.pop(100)  # Esto causaba el error porque el índice 100 no existe
Gatos.pop()      # Esto elimina correctamente el último elemento
print(Gatos)
```
