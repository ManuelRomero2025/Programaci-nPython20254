Pandas es una biblioteca de código abierto para el lenguaje Python, especializada en la manipulación y el análisis de datos. Ofrece estructuras de datos flexibles y operaciones eficientes para trabajar con datos tabulares (por ejemplo, tablas numéricas o series de tiempo), al punto que a veces se la describe como "el Excel de Python" por su facilidad para gestionar datos en formato de tabla. El nombre pandas proviene del término "panel data" (datos de panel), un concepto de econometría que se refiere a conjuntos de datos que combinan una dimensión temporal con otra dimensional transversal (por ejemplo, múltiples observaciones a lo largo del tiempo para los mismos individuos). Esta denominación también juega con las iniciales de "Python data analysis", reflejando el propósito de la biblioteca.

Pandas es una muy popular librería de código abierto dentro de los desarrolladores de Python, y sobre todo dentro del ámbito de Data Science y Machine Learning, ya que ofrece unas estructuras muy poderosas y flexibles que facilitan la manipulación y tratamiento de datos.

```python
import statistics as stat

datos = [21, 17, 89, 76, 32, 21, 45, 21, 89, 21, 15, 89, 21]

# Obtener la moda única
moda = stat.mode(datos)
print("Moda:", moda)  # Salida: 21

# Obtener múltiples modas si existen
modas = stat.multimode(datos)
print("Moda:", modas)  # Salida: [21]
```

```python
np.random.seed(1235)
panel=np.random.uniform(100, 500, size=(3,4,2))# 3 bloques, 4 filas, 2 columnas
print(panel)
```

```python
#Crear indices y etiquetas
Cliente=["Cliente 1", "Cliente 2", "Cliente 3"] #Bloques
Meses=["Enero", "Febrero","Marzo", "Abril"]# Filas
Ingresos=["Brutos", "Netos"]# columas

data=pd.DataFrame(
    panel.reshape(12, 2),# 3*4=12 bloque por fila
    index=pd.MultiIndex.from_product([Cliente,Meses],
                                     names=["Cliente", "Meses"]),
    columns=Ingresos
)
print(data)
```
