# Condicionales : if, elif, else

#if=si

```python
edad=10
if edad >= 18:
  print("Eres mayor de edad")
```

```python
a=10
b=15
if a > b:
  print(" a es mayor que b")
```

# else: si no

```python
edad=17
if edad >=18:
  print("Es mayor de edad")
else:
  print("Es menor de edad")
```

# elif: si no, entonces

```python
nota=50
if nota < 20:
  print(" Reprobado")
elif nota <= 29:
  print(" Examen supletorio")
else:
  print("Aprobado")
```

```python
#lista de clientes de su ingresos
clientes={"CO001": 15000,
          "CO002": 25000,
          "CO003": 1000,
          "CO004": 8000,
          "CO005": 2500,
          "CO006": 10000,
          "CO007": 5000,
          "CO008": 500,
          "CO009": 150000,
          "CO0010":30000,
          }
print(clientes)
#clientes.items()
```

```python
for cliente, ingreso in clientes.items():
  if ingreso < 1000:
    Categoria= "Bajo"
  elif ingreso <= 20000:
    Categoria= "Medio"
  else:
    Categoria = "Alto"
  print(f"Cliente: {cliente} - Ingreso: ${ingreso} - Categoria: {Categoria}")
```
