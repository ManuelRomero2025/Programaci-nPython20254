# 1. Series de Pandas

Una `Series` es una estructura unidimensional con valores e índice. En ciencia de datos representa una variable individual: ventas, notas, temperaturas, precios, clientes, entre otras.


```python
# Ejemplo base 1: Series de países
s = pd.Series(['Colombia', 'Argentina', 'Uruguay', 'Brasil', 'USA'], dtype='string')
print(s)
print('\nTipo de dato:')
print(s.dtype)
```

```python
# Ejemplo base 2: Series desde diccionario
s1 = pd.Series({'Colombia': 6.0, 'Argentina': 4.5, 'Uruguay': 8.5, 'Brasil': 5.0, 'USA': 5.0})
print(s1)
print('\nEstadísticos descriptivos:')
print(s1.describe())
```

## 1.1. Ventas mensuales

Representa ventas mensuales de una empresa. Permite calcular promedio, máximo, mínimo y detectar meses destacados.


```python
ventas = pd.Series([1200, 1500, 1800, 1700, 2100, 2500], index=['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio'])
print(ventas)

print('\nEstadísticos descriptivos:')
print(ventas.describe())

print('\nMayor valor:')
print(ventas.idxmax(), ventas.max())

print('\nFiltro de valores superiores al promedio:')
print(ventas[ventas > ventas.mean()])
```

## 1.2. Temperaturas semanales

Representa temperaturas registradas durante una semana. Es útil para análisis ambiental y series temporales.


```python
temperaturas = pd.Series([18, 20, 19, 22, 24, 21, 20], index=['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo'])
print(temperaturas)

print('\nEstadísticos descriptivos:')
print(temperaturas.describe())

print('\nMayor valor:')
print(temperaturas.idxmax(), temperaturas.max())

print('\nFiltro de valores superiores al promedio:')
print(temperaturas[temperaturas > temperaturas.mean()])
```

# 5. Conversión de tipos de datos con `pd.to_numeric()`

Cuando se crea un `DataFrame` con `dtype='string'`, todas las columnas quedan como texto. Para realizar cálculos estadísticos, las variables cuantitativas deben convertirse a tipo numérico.


```python
# Ejemplo base de conversión de tipos

datos = {
    'nombre': ['Cali', 'Medellin', 'Bogotá', 'Barranquilla'],
    'edad': [18, 22, 20, 21],
    'promedio': [4.5, 4.2, 3.9, 4.7],
    'grado': ['Economía', 'Medicina', 'Arquitectura', 'Economía']
}

df_conversion_base = pd.DataFrame(datos, dtype='string')
print(df_conversion_base)
print('\nTipos antes:')
print(df_conversion_base.dtypes)

df_conversion_base['edad'] = pd.to_numeric(df_conversion_base['edad'])
df_conversion_base['promedio'] = pd.to_numeric(df_conversion_base['promedio'])

print('\nTipos después:')
print(df_conversion_base.dtypes)
print('\nResumen estadístico:')
print(df_conversion_base.describe())
```

# 8. Paneles convertidos a DataFrame con MultiIndex

Pandas ya no utiliza la antigua estructura `Panel`, pero permite representar datos de panel mediante `DataFrame` con `MultiIndex`. Esta forma es muy poderosa porque facilita agrupaciones, filtros, estadísticas y transformaciones.


```python
# Ejemplo base: Panel convertido a tabla MultiIndex

clientes = ['Cliente A', 'Cliente B', 'Cliente C']
meses = ['Enero', 'Febrero', 'Marzo', 'Abril']
variables = ['Ingreso Bruto', 'Ingreso Neto']

panel_df_base = pd.DataFrame(
    panel.reshape(12, 2),
    index=pd.MultiIndex.from_product([clientes, meses], names=['Cliente', 'Mes']),
    columns=variables
)

print(panel_df_base)
print('\nPromedio por cliente:')
print(panel_df_base.groupby(level='Cliente').mean())
print('\nPromedio por mes:')
print(panel_df_base.groupby(level='Mes').mean())
print('\nTabla plana:')
print(panel_df_base.reset_index().head())
```

# 9. Resumen final de procesos estadísticos y de manipulación

| Tipo de estructura | Procesos recomendados |
|---|---|
| `Series` | `describe()`, `mean()`, `median()`, `std()`, `sort_values()`, filtros booleanos, `idxmax()`, `idxmin()` |
| `DataFrame` | `head()`, `info()`, `dtypes`, `describe()`, selección de columnas, filtros, `groupby()`, creación de columnas |
| Variables tipo texto | `value_counts()`, `astype('string')`, limpieza con `.str`, codificación para modelos |
| Variables numéricas | `pd.to_numeric()`, media, mediana, desviación estándar, normalización, detección de valores extremos |
| Matrices 2D | Promedios por fila/columna, transposición, formato largo con `melt()` |
| Paneles 3D | Promedios por ejes con NumPy, conversión a `MultiIndex`, agrupación por niveles |
| MultiIndex | `groupby(level=...)`, `reset_index()`, `melt()`, filtros por niveles, análisis longitudinal |

## Conclusión

Las estructuras de Pandas y NumPy permiten pasar de datos simples a datos complejos. Las `Series` representan variables individuales; los `DataFrame` organizan información tabular; las matrices 2D permiten simulación y análisis estructurado; y los paneles 3D, al convertirse a `MultiIndex`, facilitan el análisis de datos longitudinales, multivariados y jerárquicos.
