```python
df1.num["MonthlyIncome"].quantile(q=(0,0.25, 0.5, 0.75, 1))
```

```python
Q1 = df1.num["MonthlyIncome"].quantile(0.25)
Q3 = df1.num["MonthlyIncome"].quantile(0.75)
IQR = Q3 - Q1
LI = Q1 - 1.5 * IQR
LS = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Límite inferior:", LI)
print("Límite superior:", LS)
```

```python
outliers_income = df1.num[
    (df1.num["MonthlyIncome"] < LI) |
    (df1.num["MonthlyIncome"] > LS)
]

print("Cantidad de outliers:", len(outliers_income))
print("Porcentaje de outliers:", round(len(outliers_income) / len(df1.num) * 100, 2))
outliers_income.head()
```

```python
df1["Atipico_MonthlyIncome"] = np.where(
    (df1["MonthlyIncome"] < LI) |
    (df1["MonthlyIncome"] > LS),
    1,
    0
)
```

#Puntuación Z (Z-Score)

```python
df1.num["z_MonthlyIncome"] = zscore(df1.num["MonthlyIncome"])

outliers_z_income = df1.num[abs(df1.num["z_MonthlyIncome"]) > 2]# -3 a 3 (Atípico) , 2 a 2 (Normal)

print("Cantidad de outliers por Z-Score:", len(outliers_z_income))
outliers_z_income.head()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x=df1.num.index, y=df1.num["z_MonthlyIncome"], color="steelblue")
plt.axhline(2.5, color="red", linestyle="--")
plt.axhline(-2.5, color="red", linestyle="--")
plt.title("Valores atípicos mediante Z-Score")
plt.xlabel("Empleado")
plt.ylabel("Z-Score del ingreso mensual")
plt.show()
```
