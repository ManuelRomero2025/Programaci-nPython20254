1. Una planta de alimentos registra la temperatura en 3 salas diferentes, 4 veces al día, durante 5 días. Los datos deben organizarse en un array 3D para su análisis.


* Simula los datos con np.random.normal(loc=18, scale=2.5, size=(5, 4, 3)).

* Calcula el promedio de temperatura por sala (axis=0), por día (axis=1) y por turno (axis=2).

* Determina en qué día y turno se registró la temperatura máxima y mínima.


---



```python

np.random.seed(42)
# (días=5, turnos=4, salas=3)
temps = np.random.normal(loc=18, scale=2.5, size=(5, 4, 3))
print(temps)
```

```python
#Tomar el máximo y mínimo por (día, turno) a través de salas
dia_turno_max = temps.max(axis=2)  # (5 días × 4 turnos)
dia_turno_min = temps.min(axis=2)
print(f"Temperatura máxima por día y turno:\n{dia_turno_max}\n")
print(f"Temperatura mínima por día y turno:\n{dia_turno_min}\n")
```

```python

# Promedios solicitados
prom_por_sala  = temps.mean(axis=0)  # (turnos x salas)
prom_por_dia   = temps.mean(axis=1)  # (días x salas)
prom_por_turno = temps.mean(axis=2)  # (días x turnos)

print(f"Promedios por sala:\n{prom_por_sala}\n")
print(f"Promedios por día:\n{prom_por_dia}\n")
print(f"Promedios por turno:\n{prom_por_turno}\n")
```

4. Una empresa de análisis financiero desea evaluar el comportamiento de ingresos mensuales de un grupo de clientes. Con el fin de probar su sistema de clasificación automática, decide realizar una simulación de ingresos usando una distribución normal, cuyo valor promedio histórico es de 5,000 USD y cuya variabilidad natural ha sido estimada en 1,200 USD (desviación estándar).

La empresa quiere clasificar a los clientes en dos grupos:

🟢 Cliente Premium: ingreso mensual mayor al promedio.

⚪ Cliente Estándar: ingreso mensual menor o igual al promedio.


```python
import numpy as np

# Semilla para reproducibilidad
np.random.seed(42)

# Parámetros de la simulación
media_ingresos = 5000   # ingreso promedio
desviacion = 1200       # desviación estándar

# Simular ingresos de 10 clientes con distribución normal
ingresos = np.random.normal(loc=media_ingresos,
                            scale=desviacion,
                            size=10)

print(f"Ingreso simulado (media esperada: ${media_ingresos:.2f}, desviación: ${desviacion:.2f})\n")

# Clasificar a los clientes
for i, ingreso in enumerate(ingresos, start=1):
    if ingreso > media_ingresos:
        print(f"Cliente {i}: ${ingreso:.2f} → 🟢 Cliente Premium")
    else:
        print(f"Cliente {i}: ${ingreso:.2f} → ⚪ Cliente Estándar")
```
